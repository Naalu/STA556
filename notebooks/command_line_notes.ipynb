# Part 1: Navigation & Files

## 1️⃣ `pwd` — Print Working Directory (⭐ Essential)

**What it does**: Shows your current location in the file system.

**When to use**: When you're lost and need to know where you are.

**Syntax**: `pwd`

**Example**:

````python
!pwd
````

## 2️⃣ `ls` — List Directory Contents (⭐ Essential)

**What it does**: Shows files and folders in a directory.

**When to use**: Constantly! To see what files exist and their details.

**Common flags**:
- `-l` = long format (shows permissions, owner, size, date)
- `-a` = show hidden files (those starting with `.`)
- `-h` = human-readable sizes (KB, MB, GB)
- `-R` = recursively show subdirectories

**Popular combo**: `ls -lah` (everything you need to know)

**Example**:

````python
!ls -lah /workspaces/STA556
````

## 3️⃣ `cd` — Change Directory (⭐ Essential)

**What it does**: Navigates to a different folder.

**When to use**: When you need to move around the file system.

**Special shortcuts**:
- `cd ..` = go up one level
- `cd ~` = go to home directory
- `cd -` = go to previous directory
- `cd /` = go to root (/) directory

**Example**:

````python
!bash -c "cd /workspaces/STA556 && pwd"
````

# Part 2: File Operations

## 4️⃣ `cat` — Concatenate & Display Files (⭐ Essential)

**What it does**: Shows the contents of one or more files.

**When to use**: To quickly view file contents (best for files <100 lines).

**Options**:
- `-n` = show line numbers
- `-b` = number only non-blank lines

**Tip**: For large files, use `less` or `head`/`tail` instead.

**Example**:

````python
!cat /workspaces/STA556/README.md
````

## 5️⃣ `echo` — Print Text (⭐ Essential)

**What it does**: Outputs text to the terminal or to a file.

**When to use**: Print messages, create files with content, or debug variables.

**Common patterns**:
- `echo "text"` = print text
- `echo "text" > file.txt` = write to file (overwrites)
- `echo "text" >> file.txt` = append to file

**Example**:

````python
!echo "This is a demonstration of echo command"
!echo "Current timestamp:" && date
````

## 6️⃣ `touch` — Create Empty File or Update Timestamp (⭐ Essential)

**What it does**: Creates an empty file or updates modification time.

**When to use**: Quickly create an empty file or update timestamps.

**Syntax**: `touch filename.txt`

**Example**:

````python
!touch /tmp/demo_file.txt && ls -l /tmp/demo_file.txt
````

## 7️⃣ `mkdir` — Make Directory (⭐ Essential)

**What it does**: Creates one or more new directories.

**When to use**: Organize your files into folders.

**Best practice**: Use `-p` to create parent directories automatically

**Syntax**: `mkdir -p path/to/nested/folder`

**Example**:

````python
!mkdir -p /tmp/demo_dir/subdir && ls -R /tmp/demo_dir
````

## 8️⃣ `rm` — Remove Files or Directories (⭐ Essential) ⚠️ CAUTION

**What it does**: Deletes files or directories permanently (no trash/undo!).

**When to use**: Clean up files you no longer need.

**🚨 Important flags**:
- `-f` = force (skip confirmation)
- `-r` = recursive (delete folders and contents)
- `-i` = interactive (ask before each delete)
- `-v` = verbose (show what's being deleted)

**⚠️ PRO TIP**: Use `rm -i` to stay safe until you're comfortable with this command!

**Example**:

````python
!rm /tmp/demo_file.txt && echo "File removed successfully"
````

## 9️⃣ `cp` — Copy Files or Directories (⭐ Essential)

**What it does**: Makes a copy of files or directories.

**When to use**: Create backups or duplicate files/folders.

**Key flags**:
- `-r` = recursive (copy entire directories)
- `-i` = interactive (ask before overwriting)
- `-v` = verbose (show progress)
- `-p` = preserve (keep permissions & timestamps)

**Example**:

````python
!cp /workspaces/STA556/README.md /tmp/README_copy.md && ls -l /tmp/README_copy.md
````

## 🔟 `mv` — Move or Rename Files (⭐ Essential)

**What it does**: Moves files to a new location or renames them.

**When to use**: Organize files or rename them.

**Key flags**:
- `-i` = interactive (ask before overwriting)
- `-v` = verbose (show what's moving)
- `-n` = no-clobber (don't overwrite existing files)

**Bonus**: Works for both files AND directories!

**Example**:

````python
!mv /tmp/README_copy.md /tmp/README_renamed.md && ls -l /tmp/README_renamed.md
````

# Part 3: Text & Searching

## 1️⃣1️⃣ `grep` — Search Text Patterns (⭐⭐ Intermediate)

**What it does**: Finds lines containing a specific pattern in files.

**When to use**: Find specific content, filter data, search logs.

**Popular flags**:
- `-i` = case-insensitive
- `-n` = show line numbers
- `-c` = count matches
- `-r` = search directories recursively
- `-v` = show lines that DON'T match (invert)

**Real-world**: `grep "ERROR" logfile.txt` finds all errors

**Example**:

````python
!grep -n "STA" /workspaces/STA556/README.md
````

## 1️⃣2️⃣ `head` — Show First Lines (⭐ Essential)

**What it does**: Displays the first N lines of a file.

**When to use**: Preview file contents without opening entire file.

**Default**: Shows first 10 lines (use `-n 20` for more)

**Tip**: Great for peeking at large files before processing

**Example**:

````python
!head -n 3 /workspaces/STA556/README.md
````

## 1️⃣3️⃣ `tail` — Show Last Lines (⭐ Essential)

**What it does**: Displays the last N lines of a file.

**When to use**: Check end of files, monitor logs in real-time.

**Useful flags**:
- `-n 20` = show last 20 lines
- `-f` = follow (keep watching log file updates)

**Real-world**: `tail -f /var/log/app.log` monitors logs live!

**Example**:

````python
!tail -n 2 /workspaces/STA556/README.md
````

## 1️⃣4️⃣ `wc` — Word Count (⭐⭐ Intermediate)

**What it does**: Counts lines, words, and characters.

**When to use**: Measure file size, count data rows.

**Main flags**:
- `-l` = lines only (most useful!)
- `-w` = words only
- `-c` = bytes only
- `-m` = characters (handles multi-byte chars)

**Tip**: `wc -l *.txt` counts lines in all text files

**Example**:

````python
!wc /workspaces/STA556/README.md
!echo "---"
!wc -l /workspaces/STA556/README.md
````

## 1️⃣5️⃣ `find` — Search for Files (⭐⭐ Intermediate)

**What it does**: Searches for files/directories matching criteria.

**When to use**: Locate files you forgot about or files matching a pattern.

**Common patterns**:
- `find /path -name "*.txt"` = find all text files
- `find . -type f -size +10M` = files larger than 10MB
- `find . -mtime -7` = modified in last 7 days

**Tip**: Start with `.` to search current directory

**Example**:

````python
!find /workspaces/STA556 -name "*.md" -type f
````

## 1️⃣6️⃣ `sort` — Sort Lines (⭐⭐ Intermediate)

**What it does**: Sorts lines alphabetically or numerically.

**When to use**: Organize data, rank results, prepare for analysis.

**Key flags**:
- `-n` = numeric sort (3, 10, 100 not 10, 100, 3)
- `-r` = reverse order (Z to A)
- `-u` = unique (remove duplicates)
- `-k 2` = sort by column 2

**Real-world**: `ps aux | sort -k 3 -rn` = see most CPU-hungry processes

**Example**:

````python
!echo -e "3\n1\n2" | sort
!echo "---"
!echo -e "3\n1\n2" | sort -rn
````

**What it does**: Lists all currently running processes.

**When to use**: Find what's running, check CPU/memory usage, monitor apps.

````python
!ps aux | head -n 5
````

## 2️⃣4️⃣ `kill` — Terminate Processes (⭐⭐⭐ Advanced) ⚠️

**What it does**: Sends signals to stop processes.

**When to use**: Stop runaway programs, manage background jobs.

**Common signals**:
- `-15` = graceful terminate (default, lets app cleanup)
- `-9` = force kill (immediate, no chance to cleanup)
- `-1` = hangup (reload config)

**Workflow**: First try `kill PID`, if hangs then `kill -9 PID`

**Demo**:

````python
!echo "To terminate PID 12345: kill -9 12345"
!echo "Current shell PID:"
!echo $$
````

# Real-World Workflows

## Common Command Combinations

### 1. Find and Count
```bash
# Count lines in all Python files
find . -name "*.py" -type f | xargs wc -l | tail -1

# Count how many errors in a log
grep -c "ERROR" logfile.txt
```

### 2. Search and Replace
```bash
# Find all .txt files containing "old" and show with line numbers
grep -rn "old" .

# Count how many files contain a pattern
grep -rl "pattern" . | wc -l
```

### 3. Process Monitoring
```bash
# Show processes sorted by memory usage
ps aux | sort -k 4 -rn | head -5

# Find which process is using port 8000
ps aux | grep python
```

### 4. File Organization
```bash
# Copy and organize: create backup with timestamp
cp important.txt "important_$(date +%Y%m%d_%H%M%S).txt"

# Move old files to archive
find . -name "*.log" -mtime +30 -exec mv {} ./archive/ \;
```

### 5. Data Processing
```bash
# Sort and count unique items
cat data.txt | sort | uniq -c | sort -rn

# Extract and process CSV columns
grep "ERROR" logfile.csv | cut -d',' -f1,3 | head -10
```

# Pro Tips & Best Practices

## Safety First 🛡️
1. **Before using `rm`, test with `ls`**: Run `ls` first with same pattern to confirm what you're deleting
2. **Use `-i` flag**: Add `-i` to `rm`, `cp`, `mv` to get confirmation
3. **Read before acting**: Always check before destructive operations

## Efficiency Hacks ⚡
- **Use wildcards**: `ls *.txt` finds all text files
- **Tab completion**: Hit TAB to auto-complete long filenames
- **Arrow keys**: UP/DOWN arrows to re-run previous commands
- **Ctrl+R**: Search command history
- **Clear screen**: Type `clear` to clean up terminal

## Common Mistakes ❌
- **`rm -rf /`**: DO NOT RUN THIS! It deletes everything!
- **Forgetting `-r` with `rm`**: `rm folder` fails; use `rm -r folder`
- **Grep syntax**: `grep pattern file.txt` not `grep file.txt pattern`
- **cd without path**: `cd` goes home; `cd ..` goes up one level
- **Piping to `grep`**: `cat file | grep pattern` works but `grep pattern file` is faster

## Powerful Combinations 🚀
```bash
# Find large files
find . -type f -size +100M -exec ls -lh {} \;

# Monitor a file in real-time
tail -f logfile.log | grep ERROR

# Create a backup with timestamp
cp myfile.txt myfile_$(date +%Y%m%d).txt.bak

# Process multiple files
for file in *.txt; do wc -l "$file"; done
```

# Troubleshooting Guide

| Problem | Solution |
|---------|----------|
| "Command not found" | Use `which command` to check if installed |
| "Permission denied" | Use `chmod u+x file` to make executable |
| "File not found" | Use `ls` or `find` to locate the file |
| "No space left on device" | Use `df -h` to check, then delete old files |
| "Process won't terminate" | Try `kill PID` first, then `kill -9 PID` |
| "Can't delete file" | Check permissions with `ls -la` |
| "Stuck in command" | Press `Ctrl+C` to interrupt |
| "Wrong directory" | Use `pwd` to check, `cd` to change |
